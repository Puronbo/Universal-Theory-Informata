# Model Demo
End-to-end demonstration of the engine.

```python
from engine import net_zero, iterate, logistic, erosion_step, Playhead, w_cpl, rho_lambda_ratio, delta_norm, milestone_reached
print('net_zero([1,-1,0]):', net_zero([1,-1,0]))
x, steps, conv = iterate(logistic(2.5), 0.5)
print('logistic fixed point:', x, 'steps', steps, 'converged', conv)
s = [1,1,-1,-1,0,1,-1]
print('before erosion:', s, 'sum', sum(s))
s2 = erosion_step(s, pairs=2)
print('after erosion: ', s2, 'sum', sum(s2))
ph = Playhead(['a','b','c'], tau=1.0)
print('playhead read(1.5):', ph.read(1.5))
print('w(0.5):', w_cpl(0.5,-0.75,-0.87))
print('rho(z=1):', rho_lambda_ratio(1.0,-0.75,-0.87))
print('delta(z=1):', delta_norm(1.0,-0.75,-0.87))
print('milestone reached:', milestone_reached())
```
